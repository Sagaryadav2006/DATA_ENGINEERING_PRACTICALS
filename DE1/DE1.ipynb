### Exercise 1: File Parsers and Data Quality Checkers

```python
import csv
import json
import xml.etree.ElementTree as ET
from bs4 import BeautifulSoup
import pandas as pd
import io

def parse_and_check_csv(csv_data):
    print("--- Parsing CSV ---")
    try:
        df = pd.read_csv(io.StringIO(csv_data))
        print("Data loaded successfully:")
        display(df)

        missing_summary = df.isnull().sum()
        print("\nMissing values count per column:")
        print(missing_summary)

        duplicates = df.duplicated().sum()
        print(f"\nDuplicate rows found: {duplicates}")

        return df
    except Exception as e:
        print(f"Error parsing CSV: {e}")
        return None

def parse_and_check_json(json_data):
    print("\n--- Parsing JSON ---")
    try:
        data = json.loads(json_data)
        print("Valid JSON syntax parsed successfully.")

        if not data:
            print("Warning: The JSON document is empty.")
        else:
            print(f"Root element type: {type(data)}")
            if isinstance(data, list):
                for i, item in enumerate(data):
                    null_keys = [k for k, v in item.items() if v is None or v == ""]
                    if null_keys:
                        print(f"Warning: Record {i} has empty values in fields: {null_keys}")
        return data
    except json.JSONDecodeError as e:
        print(f"JSON Decode Error (Invalid Syntax): {e}")
        return None

def parse_and_check_html(html_data):
    print("\n--- Parsing HTML ---")
    try:
        soup = BeautifulSoup(html_data, 'html.parser')
        print("HTML parsed with BeautifulSoup.")

        tables = soup.find_all('table')
        print(f"Total tables found: {len(tables)}")
        for i, table in enumerate(tables):
            rows = table.find_all('tr')
            print(f"  Table {i+1} has {len(rows)} rows.")
            if len(rows) == 0:
                print(f"  Anomaly: Table {i+1} is empty.")
        return soup
    except Exception as e:
        print(f"Error parsing HTML: {e}")
        return None

def parse_and_check_xml(xml_data):
    print("\n--- Parsing XML ---")
    try:
        root = ET.fromstring(xml_data)
        print(f"XML root element tag: '{root.tag}' parsed successfully.")

        empty_elements = []
        for elem in root.iter():
            if elem.text is None or elem.text.strip() == "":
                if len(elem) == 0:
                    empty_elements.append(elem.tag)

        if empty_elements:
            print(f"Warning: Found empty leaf elements: {empty_elements}")
        else:
            print("No empty elements found in XML structure.")
        return root
    except ET.ParseError as e:
        print(f"XML Parse Error: {e}")
        return None

def parse_and_check_text(text_data):
    print("\n--- Parsing Plain Text ---")
    try:
        lines = text_data.strip().split('\n')
        print(f"Total lines parsed: {len(lines)}")

        empty_lines = [i+1 for i, line in enumerate(lines) if not line.strip()]
        if empty_lines:
            print(f"Warning: Found empty or blank lines at line number(s): {empty_lines}")
        return lines
    except Exception as e:
        print(f"Error parsing plain text: {e}")
        return None

csv_sample = """id,name,age,email
1,Alice,30,alice@example.com
2,Bob,,bob@example.com
3,Charlie,25,
1,Alice,30,alice@example.com"""

json_sample = """[
    {"id": 1, "name": "Product A", "price": 12.99},
    {"id": 2, "name": "Product B", "price": null},
    {"id": 3, "name": "", "price": 15.00}
]"""

html_sample = """<html>
<body>
    <h1>Product Summary</h1>
    <table>
        <tr><th>Item</th><th>Stock</th></tr>
        <tr><td>Item A</td><td>10</td></tr>
        <tr><td>Item B</td><td></td></tr>
    </table>
    <table id='empty_table'>
    </table>
</body>
</html>"""

xml_sample = """<store>
    <item>
        <id>101</id>
        <name>Laptop</name>
        <price>999</price>
    </item>
    <item>
        <id>102</id>
        <name></name>
        <price>499</price>
    </item>
</store>"""

text_sample = """First line of configuration data.
Second line contains value: 123.

Fourth line after an anomaly of an empty line.
"""

parse_and_check_csv(csv_sample)
parse_and_check_json(json_sample)
parse_and_check_html(html_sample)
parse_and_check_xml(xml_sample)
parse_and_check_text(text_sample)
```

### Exercise 2: Binary File Reader and Writer


```python
import os

def write_binary_file(filepath, byte_data):
    """Writes binary data to a file safely, creating directory structures if needed."""
    print(f"Writing to binary file: {filepath}")
    try:
        dir_name = os.path.dirname(filepath)
        if dir_name and not os.path.exists(dir_name):
            os.makedirs(dir_name)

        with open(filepath, 'wb') as binary_file:
            binary_file.write(byte_data)
        print("Binary data written successfully.")
        return True
    except IOError as e:
        print(f"I/O Error during writing binary file: {e}")
        return False
    except Exception as e:
        print(f"Unexpected error during write: {e}")
        return False

def read_binary_file(filepath):
    """Reads binary data from a file safely, returning None if errors occur."""
    print(f"\nReading from binary file: {filepath}")
    if not os.path.exists(filepath):
        print(f"Error: The file '{filepath}' does not exist.")
        return None

    try:
        with open(filepath, 'rb') as binary_file:
            data = binary_file.read()
        print(f"Successfully read {len(data)} bytes.")
        return data
    except IOError as e:
        print(f"I/O Error during reading binary file: {e}")
        return None
    except Exception as e:
        print(f"Unexpected error during read: {e}")
        return None

test_filename = "demo_file.bin"
sample_bytes = bytes([120, 3, 255, 0, 18, 99, 104, 101, 99, 107])

if write_binary_file(test_filename, sample_bytes):
    retrieved_bytes = read_binary_file(test_filename)

    if retrieved_bytes == sample_bytes:
        print("SUCCESS: Retrieved binary content matches original byte data!")
    else:
        print("FAILURE: Content mismatch.")

if os.path.exists(test_filename):
    os.remove(test_filename)
```

### Exercise 3: Pattern Matching using Regular Expressions

```python
import re

def find_patterns(pattern, text):
    """Searches for all occurrences of a pattern in the given text."""
    print(f"Searching for pattern '{pattern}'...")
    try:
        compiled_re = re.compile(pattern)
        matches = compiled_re.finditer(text)

        results = []
        for match in matches:
            results.append({
                "match": match.group(),
                "start_index": match.start(),
                "end_index": match.end()
            })

        if results:
            print(f"Found {len(results)} match(es):")
            for res in results:
                print(f" - '{res['match']}' at position {res['start_index']}:{res['end_index']}")
        else:
            print("No matches found.")
        return results
    except re.error as e:
        print(f"Invalid Regular Expression: {e}")
        return []

def split_by_pattern(pattern, text):
    """Splits a string based on match occurrences of a pattern."""
    print(f"\nSplitting text using pattern '{pattern}'...")
    try:
        segments = re.split(pattern, text)
        # Remove empty string elements from results
        segments = [seg.strip() for seg in segments if seg.strip()]
        print(f"Splits generated: {segments}")
        return segments
    except re.error as e:
        print(f"Invalid Regular Expression: {e}")
        return []

def replace_pattern(pattern, replacement, text):
    """Replaces instances of a pattern with a replacement string/callback."""
    print(f"\nReplacing pattern '{pattern}' with '{replacement}'...")
    try:
        modified_text, replacement_count = re.subn(pattern, replacement, text)
        print(f"Replacements made: {replacement_count}")
        print(f"Result: '{modified_text}'")
        return modified_text
    except re.error as e:
        print(f"Invalid Regular Expression: {e}")
        return text

target_text = "Contact us at info@example.com or support@business-hub.co.uk before tomorrow!"

find_patterns(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', target_text)

split_by_pattern(r'[\s,!.@]+', target_text)

replace_pattern(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', '[REDACTED_EMAIL]', target_text)
```
